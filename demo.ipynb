## Phase 2: LQHIII Affinity Optimization

The goal of Phase 2 is to **optimize the LQHIII toxin sequence** to improve its interaction with the cardiac-like NaV1.5 structure generated in Phase 1.

### Starting Structure

Phase 2 begins with the cardiac-like NaV1.5–LQHIII complex produced in Phase 1:

```text
7K18_AQPMSSSPKET_mutant.pdb

### Scoring Fuction DDG

````python
# amino acids to allow  mutation in LQHIII
aa_mutate = [107, 108, 109, 110, 113, 114, 117, 118, 120, 121,
                141, 142, 157, 159, 160, 161, 162, 166]

opt = AffinityOptimizer('pdb_files/7K18_AQPMSSSPKET_mutant.pdb', scoring_function='DDG', 
                        pos_to_mutate=aa_mutate, temp=20, cooling_rate=0.95, relax=False, 
                        relax_every=1, early_stop_iter=75, number_steps=2, quiet=False)
opt.run()
opt.report_score()
````

### Scoring function: Distance

````python
# amino acids to allow  mutation in LQHIII
aa_mutate = [107, 108, 109, 110, 113, 114, 117, 118, 120, 121,
                141, 142, 157, 159, 160, 161, 162, 166]

opt = AffinityOptimizer('pdb_files/7K18_AQPMSSSPKET_mutant.pdb', scoring_function='Distance', 
                        pos_to_mutate=aa_mutate, temp=20, cooling_rate=0.95, relax=False, 
                        relax_every=1, early_stop_iter=75, number_steps=2, quiet=False)

opt.insert_interaction('A', 'B', 1616, 64, 'Salt Bridge')
opt.insert_interaction('A', 'B', 1612, 15, 'Hydrogen Bond')
opt.insert_interaction('A', 'B', 1611, 43, 'Hydrogen Bond')

opt.run()
opt.report_score()
````
